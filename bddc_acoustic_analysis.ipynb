# Burmese Daily Dialogue Corpus (BDDC) — Acoustic Analysis
**Quantitative Evaluation of Acoustic Distinctiveness Across 13 Synthetic Speakers**

This notebook provides a publication-grade acoustic and prosodic analysis of the **Burmese Daily Dialogue Corpus (BDDC)**.
The objective is to empirically verify whether the 13 synthetic speaker identities are acoustically distinct across:
1. **Voice Identity Features:** Fundamental frequency ($F_0$), MFCC spectral centroids, and ECAPA-TDNN neural speaker embeddings.
2. **Speaking Style & Prosody:** Syllable speaking rates, utterance durations, and pause structures.
3. **Statistical Significance:** Kruskal-Wallis non-parametric tests and effect sizes ($\eta^2$).
